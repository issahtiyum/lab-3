### Part 1.1 — A single neuron is just a line
A neuron computes `y_hat = w * x + b`. Fit it to the sine data with **manual gradient descent**.

```python
# TODO: Generate the toy data
x = np.linspace(-3, 3, 100)
y = np.sin(x) + np.random.normal(loc = 0, scale = 0.1, size = x.shape)
# Plot it.
plt.plot(x,y)
plt.xlim(-5, 5)
plt.title("Graph of Dummy Data")
plt.show()

lr = 0.01
# TODO: Initialise w and b to small random values.
w = np.random.randn() * 0.01
b = np.random.randn() * 0.01
# TODO: Implement ONE gradient-descent step by hand, then loop it for ~200 steps:
y_hat = w * x + b
loss  = np.mean((y_hat - y) ** 2)               # MSE
dw    = np.mean(2 * (y_hat - y) * x)            # dLoss/dw  (derive this on paper!)
db    = np.mean(2 * (y_hat - y))                # dLoss/db
w    -= lr * dw
b    -= lr * db

losses = [loss]

for i in range(200):
  y_hat = w * x + b
  loss  = np.mean((y_hat - y) ** 2)               # MSE
  dw    = np.mean(2 * (y_hat - y) * x)            # dLoss/dw  (derive this on paper!)
  db    = np.mean(2 * (y_hat - y))                # dLoss/db
  w    -= lr * dw
  b    -= lr * db
  losses.append(loss)
#   Record the loss at every step.

# TODO: Plot (a) the loss curve over the 200 steps, and
#             (b) the data with the fitted line on top.

plt.plot(range(201), losses)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Graph of Loss against Epochs")
plt.show()

plt.plot(x, y_hat)
plt.plot(x,y)
plt.xlim(-5, 5)
plt.title("Graph of Dummy Data")
```

**Student Reasoning — The single neuron**
*1. Derive `dLoss/dw` on paper and include a photo or the LaTeX in this cell — does it match
the code?*
*2. The loss stops improving but the fit is clearly bad. Is this underfitting or overfitting,
and WHY can a single neuron never fit a sine wave, no matter how long you train?*

> **Answer:**  
> 1. 
> $$
>\frac{\partial loss}{\partial w} =
>\frac{\partial loss}{\partial \hat{y}}
>\cdot
>\frac{\partial \hat{y}}{\partial w}
>$$
>
>$$
>\hat{y} = wx + b
>$$
>
>$$
>\frac{\partial \hat{y}}{\partial w} = x
>$$
>
>$$
>loss = \frac{(\hat{y} - y)^2}{n}
>$$
>
>where n represents the total number of data points.
>
>$$
>\frac{\partial loss}{\partial \hat{y}} = \frac{2(\hat{y} - y)}{n}
>$$
>
>$$
>\therefore \frac{\partial loss}{\partial w} =
>\frac{2(\hat{y} - y) \cdot x}{n}
>$$
>
>For all data points,
>
>$$
>\frac{\partial loss}{\partial w} = \frac{1}{n} \sum_{i =1}^{n} 2(\hat{y}_i - y_i) \cdot x_i
>$$
>
>From the final answer, we can see that `dLoss/dw` matches the code.
>
> 2. In this case, it is underfitting. And the reason a single neuron will never fit a sine wave is because a single neuron on its own will always produce a linear hypothesis function, even though the sine function has bends and curves.

### Part 1.2 — Hidden layers and activations: why depth helps
Now build a tiny network **by hand**: 1 input -> a hidden layer of 8 neurons with `tanh` ->
1 output. Still NumPy, still manual gradients.

```python
# TODO: Initialise parameters
#   W1: shape (1, 8),  b1: shape (8,)
#   W2: shape (8, 1),  b2: shape (1,)
#   Small random values (e.g. np.random.randn(...) * 0.5)
x1=x.reshape(100,1)
y1=y.reshape(100,1)
lr=0.02
W1 = np.random.randn(1,8) * 0.5
W2 = np.random.randn(8,1) * 0.5
b1 = np.random.randn(8,) * 0.5
b2 = np.random.randn(1,) * 0.5

# TODO: Forward pass (keep the intermediate values — you need them for backprop):
#   z1 = x @ W1 + b1        # pre-activation, shape (100, 8)
#   h  = np.tanh(z1)        # activation
#   y_hat = h @ W2 + b2     # output, shape (100, 1)
#   loss  = np.mean((y_hat - y) ** 2)
z1 = x1 @ W1 + b1 
h  = np.tanh(z1)
y_hat = h @ W2 + b2
loss  = np.mean((y_hat - y1) ** 2)
losslist=[loss]


# TODO: Backward pass (the chain rule, layer by layer):
#   d_yhat = 2 * (y_hat - y) / len(y)
#   dW2 = h.T @ d_yhat ;  db2 = d_yhat.sum(axis=0)
#   dh  = d_yhat @ W2.T
#   dz1 = dh * (1 - np.tanh(z1) ** 2)     # derivative of tanh
#   dW1 = x.T @ dz1 ;  db1 = dz1.sum(axis=0)
d_yhat = 2 * (y_hat - y1) / len(y1)
dW2 = h.T @ d_yhat 
db2 = d_yhat.sum(axis=0)
dh  = d_yhat @ W2.T
dz1 = dh * (1 - np.tanh(z1) ** 2)
dW1 = x.T @ dz1 
db1 = dz1.sum(axis=0)


# TODO: Update all parameters with learning rate lr, loop for ~3000 steps,
#   record the loss.
W1=W1-(lr * dW1)
W2=W2-(lr * dW2)

for i in range(3000):
  z1 = x1 @ W1 + b1 
  h  = np.tanh(z1)
  y_hat = h @ W2 + b2
  loss  = np.mean((y_hat - y1) ** 2)
  losslist.append(loss)
  
  d_yhat = 2 * (y_hat - y1) / len(y1)
  dW2 = h.T @ d_yhat 
  db2 = d_yhat.sum(axis=0)
  dh  = d_yhat @ W2.T
  dz1 = dh * (1 - np.tanh(z1) ** 2)
  dW1 = x.T @ dz1
  db1 = dz1.sum(axis=0)

  W1=W1-(lr * dW1)
  W2=W2-(lr * dW2)
# TODO: Plot the loss curve AND the final fit over the data.


plt.plot(x1, y_hat)
plt.plot(x1,y1)
plt.title("Graph of hypothetical function and actual function with activation function")
plt.show()

plt.plot(range(3001), losslist)
plt.title("Graph of loss values against number of epochs")
plt.show()



#   Then re-run with the tanh REMOVED (h = z1). What happens to the fit?
W1 = np.random.randn(1,8) * 0.5
W2 = np.random.randn(8,1) * 0.5
b1 = np.random.randn(8,) * 0.5
b2 = np.random.randn(1,) * 0.5

z1 = x1 @ W1 + b1 
h = z1
y_hat = h @ W2 + b2
loss  = np.mean((y_hat - y1) ** 2)
losslist=[loss]

d_yhat = 2 * (y_hat - y1) / len(y1)
dW2 = h.T @ d_yhat 
db2 = d_yhat.sum(axis=0)
dh  = d_yhat @ W2.T
dz1 = dh * (1 - np.tanh(z1) ** 2)
dW1 = x.T @ dz1 
db1 = dz1.sum(axis=0)


W1=W1-(lr * dW1)
W2=W2-(lr * dW2)

for i in range(3000):
  z1 = x1 @ W1 + b1 
  h = z1
  y_hat = h @ W2 + b2
  loss  = np.mean((y_hat - y1) ** 2)
  losslist.append(loss)
  
  d_yhat = 2 * (y_hat - y1) / len(y1)
  dW2 = h.T @ d_yhat 
  db2 = d_yhat.sum(axis=0)
  dh  = d_yhat @ W2.T
  dz1 = dh * (1 - np.tanh(z1) ** 2)
  dW1 = x.T @ dz1
  db1 = dz1.sum(axis=0)

  W1=W1-(lr * dW1)
  W2=W2-(lr * dW2)



plt.plot(x1, y_hat)
plt.plot(x1,y1)
plt.title("Graph of hypothetical function and actual function without activation function")
plt.show()

plt.plot(range(3001), losslist)
plt.title("Graph of loss values against number of epochs")
plt.show()
```

*2. What role did the hidden layer's 8 neurons play in fitting the curve? (Hint: plot a few

*3. This is exactly what `loss.backward()` will automate in Section 2. In one sentence, what

> **Answer:**
> 1. $$ z1 =  x \cdot W_1 + b_1 $$
>
>With activation function,$$ h = tanh(z1) $$
>$$ \implies h = tanh(x \cdot W_1 + b_1) $$
>$$ \hat{y} = h \cdot W_2 + b_2 $$
>$$ \implies \hat{y} = tanh(x \cdot W_1 + b_1) \cdot W_2 + b_2 $$
>
>This output produced is not a linear function. Therefore layers stacked on top of each other do not produce a linear function.
>
>
>Without an activation function, 
>$$ h = z1 $$
>$$ \implies h = x \cdot W_1 + b_1 $$
>$$ \hat{y} = h \cdot W_2 + b_2 $$
>$$ \implies \hat{y} = (x \cdot W_1 + b_1) \cdot W_2 + b_2 $$
>
>This output produces a linear function. The layers stacked on top of each other collapse to form a linear function, no matter the matter of neurons or layers used.

### Part 1.3 — The learning rate: too cold, too hot, just right
The single most important hyperparameter. See its three regimes with your own eyes.

```python
# TODO: Wrap your Part 1.2 training loop in a function train_toy(lr, steps=3000)
#   that returns the list of losses.
np.random.seed(RANDOM_STATE)

def train_toy(lr, steps=3000):
  x1=x.reshape(100,1)
  y1=y.reshape(100,1)
  W1 = np.random.randn(1,8) * 0.5
  W2 = np.random.randn(8,1) * 0.5
  b1 = np.random.randn(8,) * 0.5
  b2 = np.random.randn(1,) * 0.5
  losslist = []

  for i in range(steps):
    z1 = x1 @ W1 + b1 
    h  = np.tanh(z1)
    y_hat = h @ W2 + b2
    loss  = np.mean((y_hat - y1) ** 2)
    losslist.append(loss)
    
    d_yhat = 2 * (y_hat - y1) / len(y1)
    dW2 = h.T @ d_yhat 
    db2 = d_yhat.sum(axis=0)
    dh  = d_yhat @ W2.T
    dz1 = dh * (1 - np.tanh(z1) ** 2)
    dW1 = x.T @ dz1
    db1 = dz1.sum(axis=0)

    W1=W1-(lr * dW1)
    W2=W2-(lr * dW2)
  return losslist


# TODO: Run it with THREE learning rates, e.g. lr = 0.001, 0.05, 1.0
#   (keep the same initial weights for a fair comparison — reset the seed).
learning_rates = [0.001, 0.05, 1.0]
losslists = []
for lr in learning_rates:
  losslists.append(train_toy(lr))


# TODO: Plot the three loss curves on the same axes (log scale on y helps), with a legend.
for i in range(len(learning_rates)):
  plt.plot(range(3000), np.log(losslists[i]), label = f"learning rate = {learning_rates[i]}")
plt.legend()
plt.ylim(-5,20)
plt.show()
```

**Student Reasoning — Learning rate**
*Describe what each of the three curves shows. Why does a too-large learning rate make the
loss explode or oscillate instead of just "learning faster"?*

> **Answer:** For learning rates of 0.001 and 0.05, the values of the losses quickly decrease at the early stages, but slowly decrease after some time, creating an elbow shape. For the learning rate of 1.0, the loss values rapidly explode as the number of iterations increase. The reason the loss explode or oscillate is due to the fact that as the weight goes down the curve, the learning weight is too large for it to reach the bottom, so it just jumps back and forth over the target.
