### Part 1.1 — A single neuron is just a line
A neuron computes `y_hat = w * x + b`. Fit it to the sine data with **manual gradient descent**.

```python
# TODO: Generate the toy data
np.random.seed(RANDOM_STATE)
x = np.linspace(-3, 3, 100)
y = np.sin(x) + np.random.normal(loc = 0, scale = 0.1, size = x.shape)
# Plot it.
plt.plot(x,y)
plt.xlim(-5, 5)
plt.title("Graph of Dummy Data")
plt.show()

lr = 0.01
# TODO: Initialise w and b to small random values.
w = np.random.randn() * 0.01
b = np.random.randn() * 0.01
# TODO: Implement ONE gradient-descent step by hand, then loop it for ~200 steps:
y_hat = w * x + b
loss  = np.mean((y_hat - y) ** 2)               # MSE
dw    = np.mean(2 * (y_hat - y) * x)            # dLoss/dw  (derive this on paper!)
db    = np.mean(2 * (y_hat - y))                # dLoss/db
w    -= lr * dw
b    -= lr * db

losses = [loss]

for i in range(200):
  y_hat = w * x + b
  loss  = np.mean((y_hat - y) ** 2)               # MSE
  dw    = np.mean(2 * (y_hat - y) * x)            # dLoss/dw  (derive this on paper!)
  db    = np.mean(2 * (y_hat - y))                # dLoss/db
  w    -= lr * dw
  b    -= lr * db
  losses.append(loss)
#   Record the loss at every step.

# TODO: Plot (a) the loss curve over the 200 steps, and
#             (b) the data with the fitted line on top.

plt.plot(range(201), losses)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Graph of Loss against Epochs")
plt.show()

plt.plot(x, y_hat)
plt.plot(x,y)
plt.xlim(-5, 5)
plt.title("Graph of Dummy Data")
```

### Part 1.3 — The learning rate: too cold, too hot, just right
The single most important hyperparameter. See its three regimes with your own eyes.

```python
# TODO: Wrap your Part 1.2 training loop in a function train_toy(lr, steps=3000)
#   that returns the list of losses.
np.random.seed(RANDOM_STATE)

def train_toy(lr, steps=3000):
  x1=x.reshape(100,1)
  y1=y.reshape(100,1)
  W1 = np.random.randn(1,8) * 0.5
  W2 = np.random.randn(8,1) * 0.5
  b1 = np.random.randn(8,) * 0.5
  b2 = np.random.randn(1,) * 0.5
  losslist = []

  for i in range(steps):
    z1 = x1 @ W1 + b1 
    h  = np.tanh(z1)
    y_hat = h @ W2 + b2
    loss  = np.mean((y_hat - y1) ** 2)
    losslist.append(loss)
    
    d_yhat = 2 * (y_hat - y1) / len(y1)
    dW2 = h.T @ d_yhat 
    db2 = d_yhat.sum(axis=0)
    dh  = d_yhat @ W2.T
    dz1 = dh * (1 - np.tanh(z1) ** 2)
    dW1 = x1.T @ dz1
    db1 = dz1.sum(axis=0)

    W1=W1-(lr * dW1)
    W2=W2-(lr * dW2)
  return losslist


# TODO: Run it with THREE learning rates, e.g. lr = 0.001, 0.05, 1.0
#   (keep the same initial weights for a fair comparison — reset the seed).
learning_rates = [0.001, 0.05, 1.0]
losslists = []
for lr in learning_rates:
  losslists.append(train_toy(lr))


# TODO: Plot the three loss curves on the same axes (log scale on y helps), with a legend.
for i in range(len(learning_rates)):
  plt.plot(range(3000), np.log(losslists[i]), label = f"learning rate = {learning_rates[i]}")
plt.legend()
plt.ylim(-5,20)
plt.show()
```

> **Answer:** For learning rates of 0.001 and 0.05, the values of the losses quickly decrease at the early stages, but slowly decrease after some time, creating an elbow shape. For the learning rate of 1.0, the loss values rapidly explode as the number of iterations increase. ### Part 2.1 — Preprocess and split (recycled from Lab 2)
Reuse your Lab 2 pipeline: encode categoricals, scale numerics, stratified three-way split.

```python
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder, StandardScaler
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split

# TODO: Load the obesity dataset (same OBESITY_URL as Lab 2).
OBESITY_URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/GkDzb7bWrtvGXdPOfk6CIg/Obesity-level-prediction-dataset.csv"

obesity = pd.read_csv(OBESITY_URL)


# TODO: Encode categoricals exactly as you did in Lab 2
#   (binary yes/no -> 0/1; Gender/MTRANS/CAEC/CALC -> your justified choice).
#   Encode the target NObeyesdad into integers 0..6 (LabelEncoder) — CrossEntropyLoss
#   expects integer class indices.

decision = {"yes": 1, "no": 0}

obesity["FAVC"] = obesity["FAVC"].map(decision)
obesity["SMOKE"] = obesity["SMOKE"].map(decision)
obesity["SCC"] = obesity["SCC"].map(decision)
obesity["family_history_with_overweight"] = obesity["family_history_with_overweight"].map(decision)


ord_encoder = OrdinalEncoder(categories=[
    ["no", "Sometimes", "Frequently", "Always"]
])

obesity[["CAEC"]] = ord_encoder.fit_transform(obesity[["CAEC"]])
obesity[["CALC"]] = ord_encoder.fit_transform(obesity[["CALC"]])

obesity = pd.get_dummies(obesity, columns=["Gender", "MTRANS"], drop_first=True)

obesity["BMI"] = obesity["Weight"] / (obesity["Height"]**2)

# classes = obesity['NObeyesdad'].unique()
label_encoder = LabelEncoder()
obesity["NObeyesdad"] = label_encoder.fit_transform(obesity["NObeyesdad"])

X = obesity.drop(columns = "NObeyesdad")
#Same as X = obesity.drop("NObeyesdad", axis = 1)
y = obesity["NObeyesdad"]

# TODO: Stratified train / validation / test split (60/20/20, stratify=y,
#   random_state=RANDOM_STATE).
X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=0.25, random_state=RANDOM_STATE, stratify = y_temp
)

# TODO: Fit a StandardScaler on the TRAINING set only; transform all three splits.
#   (Neural networks are far more sensitive to feature scale than random forests — 
#    remember Section 1, where all gradients depended on x!)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_val_scaled = scaler.transform(X_val)

# TODO: Convert to tensors and wrap in DataLoaders:
#   X -> torch.float32, y -> torch.long
#   train_loader (batch_size=32, shuffle=True), val_loader, test_loader
X_train_tensor = torch.tensor(X_train_scaled, dtype=torch.float32)
X_val_tensor = torch.tensor(X_val_scaled, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test_scaled, dtype=torch.float32)

y_train_tensor = torch.tensor(y_train.to_numpy(), dtype=torch.long)
y_val_tensor = torch.tensor(y_val.to_numpy(), dtype=torch.long)
y_test_tensor = torch.tensor(y_test.to_numpy(), dtype=torch.long)

train_loader = DataLoader(
  TensorDataset(X_train_tensor, y_train_tensor), 
  batch_size=32, 
  shuffle=True
)

val_loader = DataLoader(
  TensorDataset(X_val_tensor, y_val_tensor), 
  batch_size=32, 
  shuffle=True
)

test_loader = DataLoader(
  TensorDataset(X_test_tensor, y_test_tensor), 
  batch_size=32, 
  shuffle=True
)

```

### Part 2.2 — Design the network
An MLP: input -> hidden layers with ReLU -> 7 output logits.

```python
import torch.nn.functional as F
# TODO: Define a FeedForwardNet class (subclass nn.Module) with:
#   - constructor args: input_dim, hidden_sizes=(64, 32), n_classes=7
#   - at least TWO hidden layers with ReLU activations
#   - output layer producing 7 raw logits
#     (NO softmax — nn.CrossEntropyLoss applies log-softmax internally)

class FeedForwardNet(nn.Module):
  # def __init__(self, input_dim, hidden_sizes=(64, 32), n_classes=7):
  #   super().__init__()

  #   self.fc1 = nn.Linear(input_dim, hidden_sizes[0])
  #   self.fc2 = nn.Linear(hidden_sizes[0], hidden_sizes[1])
  #   self.output = nn.Linear(hidden_sizes[1], n_classes)


  # def forward(self, x):
  #   x = F.relu(self.fc1(x))
  #   x = F.relu(self.fc2(x))
  #   x = self.output(x)

  #   return x

  def __init__(self, input_dim, hidden_sizes=(64, 32), n_classes=7, dropout = 0.0):
    super().__init__()

    self.hidden_sizes = hidden_sizes
    self.layers = nn.ModuleDict()
    self.dropout = nn.Dropout(dropout)

    in_dim = input_dim

    for i in range(len(hidden_sizes)):
        self.layers[f"fc{i + 1}"] = nn.Linear(in_dim, hidden_sizes[i])
        in_dim = hidden_sizes[i]
          

    self.output = nn.Linear(in_dim, n_classes)

  def forward(self, x):
    for i in range(len(self.hidden_sizes)):
      x = F.relu(self.layers[f"fc{i + 1}"](x))
      x = self.dropout(x)

    x = self.output(x)

    return x
        

# TODO: Instantiate, move to DEVICE, print the model.
torch.manual_seed(RANDOM_STATE)
ffn_model = FeedForwardNet(X_train.shape[1])

ffn_model.to(DEVICE)

print(ffn_model)

# TODO: Count trainable parameters:
#   sum(p.numel() for p in model.parameters() if p.requires_grad)
trainable_params = sum(p.numel() for p in ffn_model.parameters() if p.requires_grad)
print("Trainable parameters: ", trainable_params)

#   Verify the count for the FIRST layer by hand: input_dim * 64 + 64.
```

### Part 2.3 — The training loop
The same loop you built by hand in Section 1 — now with autograd, an optimizer, and batches.

```python
import time
# TODO: criterion = nn.CrossEntropyLoss()
#       optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)   # or SGD — justify
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(ffn_model.parameters(), lr=1e-3)   # or SGD — justify

epochs = 50
best_val_accuracy = 0
history = {"train_losses": [], 
           "train_accuracies": [],
           "val_losses": [], 
           "val_accuracies": []
          }
best_model_time = 0
ffn_start = time.time()

# TODO: Train for at least 50 epochs. Each epoch:
for epoch in range(epochs):
    ffn_model.train()

    train_loss = 0
    train_correct = 0
    train_total = 0

    for xb, yb in train_loader:
        optimizer.zero_grad()
        logits = ffn_model(xb.to(DEVICE))
        loss = criterion(logits, yb.to(DEVICE))
        loss.backward()          # <- the backprop you wrote by hand in Part 1.2
        optimizer.step()        # <- the update rule you wrote in Part 1.1

        train_loss += loss.item() * xb.size(0)
        train_correct += (logits.argmax(dim=1) == yb.to(DEVICE)).sum().item()
        train_total += yb.size(0)

    train_loss /= train_total
    train_accuracy = train_correct / train_total
    history["train_losses"].append(train_loss)
    history["train_accuracies"].append(train_accuracy)

#   2. model.eval() + torch.no_grad(): compute loss AND accuracy on train and validation sets
    ffn_model.eval()
    with torch.no_grad():

      val_loss = 0
      val_correct = 0
      val_total = 0
      for xb, yb in val_loader:
          logits = ffn_model(xb.to(DEVICE))
          loss = criterion(logits, yb.to(DEVICE))
                                               
          val_loss += loss.item() * xb.size(0)
          val_correct += (logits.argmax(dim=1) == yb.to(DEVICE)).sum().item()
          val_total += yb.size(0)

    val_loss /= val_total
    val_accuracy = val_correct / val_total
    history["val_losses"].append(val_loss)
    history["val_accuracies"].append(val_accuracy)

#   3. Append everything to history lists; print a one-line summary per epoch.
    print(f'Epoch No: {epoch + 1}, Training Accuracy: {train_accuracy}, Training Loss: {train_loss}, Validation Accuracy: {val_accuracy}, Validation Loss: {val_loss}')


# TODO: Save the best model (highest validation accuracy):
    if val_accuracy > best_val_accuracy:
        best_val_accuracy = val_accuracy
        ffn_end = time.time()
        ffn_train_time = ffn_end - ffn_start
        torch.save(ffn_model.state_dict(), "ffn_best.pt")

# TODO: Plot loss curves (train + val) and accuracy curves (train + val), with legends.
plt.plot(range(epochs), history["train_losses"], label = "Training")
plt.plot(range(epochs), history["val_losses"], label = "Validation")
plt.legend()
plt.title("Loss Curves")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()

plt.plot(range(epochs), history["train_accuracies"], label = "Training")
plt.plot(range(epochs), history["val_accuracies"], label = "Validation")
plt.legend()
plt.title("Accuracy Curves")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.show()
```
